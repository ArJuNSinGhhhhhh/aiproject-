```python
# Lab Assignment 4 - copy each CELL block into a separate Colab cell (top to bottom)


# # Lab Assignment 4 — Logistic Regression (Breast Cancer Wisconsin Diagnostic)
# **CSET343 – AI in Healthcare**
#
# Run all cells top to bottom (Runtime → Run all). No download needed: the dataset comes built-in with scikit-learn
# (it is the same Wisconsin Diagnostic data: 569 patients, 30 features).


# ===== CELL 1 =====
import warnings; warnings.filterwarnings("ignore")
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             classification_report, confusion_matrix, ConfusionMatrixDisplay,
                             roc_curve, roc_auc_score)
sns.set_style("whitegrid")


# ## Task 1: Data Acquisition and Exploration


# ===== CELL 2 =====
# 1. Load the dataset
raw = load_breast_cancer(as_frame=True)
df = raw.frame.copy()

# sklearn codes 0 = malignant, 1 = benign. We flip it so that 1 = Malignant (the "positive" / dangerous class)
df["Outcome"] = 1 - df["target"]
df = df.drop(columns="target")

feature_cols = [c for c in df.columns if c != "Outcome"]
print("Shape:", df.shape)
df.head()


# ===== CELL 3 =====
# 2. Summary statistics (mean, median, std, min, max ...)
summary = df[feature_cols].describe().T
summary["median"] = df[feature_cols].median()
summary["skew"] = df[feature_cols].skew()
summary.round(3)


# ===== CELL 4 =====
# 3a. Missing values and duplicates
print("Missing values :", df.isna().sum().sum())
print("Duplicate rows :", df.duplicated().sum())
df = df.drop_duplicates()

# 3b. Outliers with the IQR rule
Q1, Q3 = df[feature_cols].quantile(0.25), df[feature_cols].quantile(0.75)
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
is_out = (df[feature_cols] < lower) | (df[feature_cols] > upper)
print("\nTotal outlier cells:", int(is_out.sum().sum()))
print("Features with most outliers:")
print(is_out.sum().sort_values(ascending=False).head(6))

# Handling: CAP (winsorize) instead of deleting rows -> we keep all 569 patients (dataset is small)
df[feature_cols] = df[feature_cols].clip(lower=lower, upper=upper, axis=1)
print("\nAfter capping, outlier cells:", int(((df[feature_cols] < lower) | (df[feature_cols] > upper)).sum().sum()))


# ===== CELL 5 =====
# 4a. Histograms + boxplots of a few key features (split by class)
key = ["mean radius", "mean texture", "mean area", "mean concavity"]
fig, ax = plt.subplots(2, 4, figsize=(17, 7))
for i, c in enumerate(key):
    sns.histplot(data=df, x=c, hue="Outcome", kde=True, ax=ax[0, i])
    sns.boxplot(data=df, x="Outcome", y=c, ax=ax[1, i])
plt.tight_layout(); plt.show()


# ===== CELL 6 =====
# 4b. Correlation heatmap
plt.figure(figsize=(14, 11))
sns.heatmap(df.corr(), cmap="coolwarm", center=0)
plt.title("Correlation heatmap"); plt.show()

print("Top features correlated with Outcome (Malignant):")
print(df.corr()["Outcome"].drop("Outcome").abs().sort_values(ascending=False).head(8).round(3))


# ===== CELL 7 =====
# 4c + 5. Class distribution
counts = df["Outcome"].value_counts().sort_index()
print(counts.rename({0: "Benign (0)", 1: "Malignant (1)"}))
print((counts / len(df) * 100).round(1).rename({0: "Benign %", 1: "Malignant %"}))

sns.countplot(x="Outcome", data=df)
plt.xticks([0, 1], ["Benign (0)", "Malignant (1)"]); plt.title("Class distribution"); plt.show()


# **Class imbalance discussion:** about 63% Benign vs 37% Malignant — a *mild* imbalance, not severe.
# Implications: (1) plain **accuracy can look good** even if the model misses many malignant cases, so we must also check
# **recall / F1 / AUC**; (2) the model may lean a little towards the majority class; (3) we must use a **stratified split**
# so train and test keep the same ratio. If it were more severe we could use `class_weight="balanced"` or oversampling (SMOTE).


# ## Task 2: Data Preprocessing


# ===== CELL 8 =====
# Target is already binary (0 = Benign, 1 = Malignant)
X = df[feature_cols]
y = df["Outcome"]

# 80/20 stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
print("Train:", X_train.shape, "| Test:", X_test.shape)
print("Malignant share -> train: %.3f, test: %.3f" % (y_train.mean(), y_test.mean()))

# Standardize (fit on TRAIN only, then apply to test -> no data leakage)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)


# ## Task 3: Model Training and Evaluation


# ===== CELL 9 =====
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_s, y_train)

y_pred = model.predict(X_test_s)
y_prob = model.predict_proba(X_test_s)[:, 1]      # probability of Malignant

print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("\nClassification report (Precision / Recall / F1 for both classes):\n")
print(classification_report(y_test, y_pred, target_names=["Benign (0)", "Malignant (1)"]))


# ===== CELL 10 =====
# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm, display_labels=["Benign", "Malignant"]).plot(cmap="Blues")
plt.title("Confusion Matrix"); plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"True Negatives  (benign, predicted benign)       : {tn}")
print(f"False Positives (benign, predicted malignant)    : {fp}  -> extra tests / anxiety")
print(f"False Negatives (malignant, predicted benign)    : {fn}  -> MISSED CANCER (most dangerous)")
print(f"True Positives  (malignant, predicted malignant) : {tp}")


# ===== CELL 11 =====
# ROC curve + AUC
fpr, tpr, _ = roc_curve(y_test, y_prob)
auc = roc_auc_score(y_test, y_prob)
plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, lw=2, label=f"Logistic Regression (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], "k--", label="Random guess")
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curve"); plt.legend(); plt.show()


# ### Interpretation (medical context)
# - **Accuracy** tells the overall share of correct predictions, but in diagnosis it is not enough on its own.
# - **Recall (sensitivity) of the Malignant class** is the most important number: a *False Negative* means a cancer patient is
#   sent home as healthy, which is far more costly than a *False Positive* (an extra biopsy).
# - **Precision of the Malignant class** tells how many of the "malignant" alarms were real — low precision means unnecessary procedures.
# - **F1-score** balances precision and recall.
# - **AUC close to 1** means the model separates malignant from benign very well across *all* thresholds. Since a missed cancer is
#   costlier, in practice we could lower the threshold (e.g. 0.5 → 0.3) to boost recall at the cost of a few more false alarms.
# - The model is a decision-support tool; a doctor must confirm the final diagnosis.


# ## Bonus: Regularization and Model Interpretation


# ===== CELL 12 =====
# Regularization: smaller C = stronger penalty. L1 also sets some coefficients to exactly 0 (feature selection).
rows = []
for penalty in ["l1", "l2"]:
    for C in [0.001, 0.01, 0.1, 1, 10, 100]:
        m = LogisticRegression(penalty=penalty, C=C, solver="liblinear", max_iter=1000).fit(X_train_s, y_train)
        p = m.predict(X_test_s)
        rows.append({"penalty": penalty, "C": C,
                     "accuracy": accuracy_score(y_test, p),
                     "recall_malignant": recall_score(y_test, p),
                     "AUC": roc_auc_score(y_test, m.predict_proba(X_test_s)[:, 1]),
                     "non-zero coefs": int((m.coef_ != 0).sum())})
pd.DataFrame(rows).round(3)


# ===== CELL 13 =====
# Interpretation: coefficients (features were standardized, so sizes are comparable)
coef = pd.Series(model.coef_[0], index=feature_cols).sort_values()
top = pd.concat([coef.head(7), coef.tail(7)])
top.plot(kind="barh", figsize=(8, 6), color=np.where(top > 0, "crimson", "steelblue"))
plt.title("Top coefficients (red = pushes towards Malignant, blue = towards Benign)")
plt.axvline(0, color="k"); plt.tight_layout(); plt.show()

# Odds ratio: how the odds of malignancy change for +1 standard deviation of the feature
odds = np.exp(coef).sort_values(ascending=False)
print("Odds ratios (top 5 risk factors):")
print(odds.head(5).round(2))
```
